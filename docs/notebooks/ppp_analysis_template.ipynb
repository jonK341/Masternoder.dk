# PPP Analysis Template — MasterNoder profit daemon

Starter notebook for Profit Path Protocol (PPP) research. Fetch redacted exports via API or read local JSONL.

```python
import json
from pathlib import Path

BASE = Path('..').resolve().parent
LEDGER = BASE / 'data' / 'crypto_exchange' / 'profit_path_ledger.jsonl'
rows = []
if LEDGER.exists():
    for line in LEDGER.read_text(encoding='utf-8').splitlines():
        if line.strip():
            rows.append(json.loads(line))
print(f'Loaded {len(rows)} PPP rows')
```

```python
from collections import Counter
skips = Counter(r.get('skip_reason') or 'unknown' for r in rows if r.get('decision') == 'skip')
fills = sum(1 for r in rows if r.get('decision') == 'execute')
print('Fills:', fills)
print('Top skips:', skips.most_common(8))
```
